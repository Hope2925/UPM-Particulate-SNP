# Calculate the p-value cutoffs to use for TF motifs based on how common they are in the genome set we used
This is based on [Townsend et al 2025](https://www.biorxiv.org/content/10.1101/2025.09.12.675852v1)

```r
library(data.table)
library(stringr)
library(dplyr)
```

```r
res <- fread("/scratch/Users/hoto7260/Resp_Env//Comb_UPM_WSP_ADP/ADP/TFEA/out/NonTSS_ADP_par_rat_Wd_HK_noback/results.txt")
res[1:2,]
colnames(res) <- c("TF", "Escore", "CorrEscore", "Events", "GC", "FPKM", "Padj", "CorrPadj", "MB_LE", "MB_LE_stdev", "SE_LE", "SE_LE_stdev", "Frac", "Na")
res[1:10,]

gene_res <- fread("/scratch/Users/hoto7260/Resp_Env//Comb_UPM_WSP_ADP/ADP/TFEA/out/Gene_ADP_par_rat_Wd_HK_noback/results.txt")

colnames(gene_res) <- c("TF", "Escore", "CorrEscore", "Events", "GC", "FPKM", "Padj", "CorrPadj", "MB_LE", "MB_LE_stdev", "SE_LE", "SE_LE_stdev", "Frac", "Na")
gene_res[1:10,]
```

```r
dim(res[res$Events > 5000,])
quantile(res[res$Events > 5000,]$Events)
dim(res[res$Events < 700 & res$Events > 0,])
quantile(res[res$Events < 700 & res$Events > 0,]$Events)

too_low <- res[res$Events < 700 & res$Events > 0,]$TF
too_high <- res[res$Events > 5000 ,]$TF
length(too_low)
length(too_high) 


```

```r
dim(gene_res[gene_res$Events > 7000,])
quantile(gene_res[gene_res$Events > 7000,]$Events)
dim(gene_res[gene_res$Events < 700 & gene_res$Events > 0,])
quantile(gene_res[gene_res$Events < 700 & gene_res$Events > 0,]$Events)

gene_too_low <- gene_res[gene_res$Events < 700 & gene_res$Events > 0,]$TF
gene_too_high <- gene_res[gene_res$Events > 5000 ,]$TF
length(gene_too_low)
length(gene_too_high) 
```

```r
setdiff(gene_too_high, too_high)
only_gene_too_low = setdiff(gene_too_low, too_low)
length(only_gene_too_low)
quantile(gene_res[gene_res$TF %in% only_gene_too_low,]$Events)
```

```r
quantile(gene_res[gene_res$TF %in% too_high,]$Events)
quantile(gene_res[gene_res$TF %in% too_low,]$Events)
```

```r
# read in pvalue
pval_file <- fread("/scratch/Users/hoto7260/assets/PVAL_CUTOFF_ESTIMATES_H12.CORE.CUWA.txt")
pval_file[1:3,]
table(pval_file$best_pval)
```

```r
table(pval_file[pval_file$MOTIF %in% too_low,]$best_pval)
table(pval_file[pval_file$MOTIF %in% too_high,]$best_pval)
```

```r
get_less_string_dict <- c("5e-11"="1e-10", "5e-10"="1e-9", "5e-09"="1e-8", "1e-08"="5e-8", "5e-08"="1e-7", "1e-07"="5e-7", 
                          "5e-07"="1e-6", "1e-06"="5e-6", "5e-06"="1e-5", "1e-05"="5e-5", "5e-05"="1e-4")
get_more_string_dict <- c("1e-11"="5e-12", "5e-11"="1e-11", "5e-10"="1e-10", "1e-09"="5e-10", "5e-09"="1e-9", "1e-08"="5e-9", "5e-08"="1e-8", "1e-07"="5e-8", 
                          "5e-07"="1e-7", "1e-06"="5e-7", "5e-06"="1e-6", "1e-05"="5e-6", "5e-05"="1e-5")
# # Convert named vector to numeric
# get_less_string_dict <- setNames(as.numeric(names(get_less_string_dict)), as.numeric(get_less_string_dict))
# get_more_string_dict <- setNames(as.numeric(names(get_more_string_dict)), as.numeric(get_more_string_dict))

pval_file <- pval_file %>%
  mutate(new_best_pval = ifelse(MOTIF %in% too_high, 
                            get_more_string_dict[as.character(best_pval)], best_pval)) %>%
  mutate(new_best_pval = ifelse(MOTIF %in% too_low, 
                            get_less_string_dict[as.character(best_pval)], new_best_pval))


pval_file$new_best_pval <- as.numeric(pval_file$new_best_pval)

pval_file[1:3,]
```

```r
as.character(pval_file$best_pval[1])
get_more_string_dict[as.character(pval_file$best_pval[1])]
```

```r
pval_file$diff = pval_file$new_best_pval - pval_file$best_pval
pval_file[pval_file$MOTIF %in% too_high,]
```

```r

pval_file[1:2,]
quantile(pval_file$diff)
quantile(pval_file[pval_file$MOTIF %in% too_high,]$diff)
quantile(pval_file[pval_file$MOTIF %in% too_low,]$diff)
```

```r
pval_file[is.na(pval_file$diff),]
```

```r
new_pval_file <- pval_file[,c("MOTIF", "new_best_pval")]
colnames(new_pval_file) <- c("MOTIF", "best_pval")

write.table(new_pval_file, 
            "/scratch/Users/hoto7260/assets/PVAL_CUTOFF_ESTIMATES_H12_noback.CORE.CUWA.txt", 
           quote=FALSE, row.names=FALSE, sep="\t")
```
